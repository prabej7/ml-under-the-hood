# **Newton's Method**

$$
x_{k+1} = x_k - \frac{g'(x_k)}{g''(x_k)}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plot(fn, title=None, xlabel=None, ylabel=None):
    xs = np.linspace(0.1, 2, 1000)
    ys = fn(xs)
    
    plt.grid()
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.plot(xs, ys)
```

### Given a function:
$$
g(x) = e^x - log(x)
$$

```python
g  = lambda x: np.exp(x) - np.log(x)
```

### We calculate $g'(x)$, we get:
$$
g'(x) = e^x - \frac{1}{x}
$$

```python
g_prime = lambda x: np.exp(x) - (1 / x)
```

### Then the second derivative:
- To know the rate of change of the slope

$$
(g'(x))' = e^x + \frac{1}{x^2}
$$

```python
g_prime_prime = lambda x: np.exp(x) + (1 / np.square(x))
```

### Newton's Method

$$
x_{k+1} = x_k - \frac{g'(x_k)}{g''(x_k)}
$$

```python
def newtons_method(x0, g_prime, g_prime_prime):
    return x0 - (g_prime(x0) / g_prime_prime(x0))
```

### **Iterations**

```python
x = 0.05
history = [x]
for _ in range(1000):
    new_x = newtons_method(x, g_prime, g_prime_prime)
    history.append(new_x)
    if abs(new_x - x) < 1e-6:
        break

    x = new_x
```

### **Plotting the result**

```python
xs = np.linspace(0.05, 1.5, 1000)

plt.plot(xs, g(xs), label="g(x)")

ys = g(np.array(history))
plt.plot(history, ys, marker="o", label="Newton iterations")

plt.xlabel("x")
plt.ylabel("g(x)")
plt.title("Newton's Method")
plt.grid()
plt.legend()
plt.show()
```
