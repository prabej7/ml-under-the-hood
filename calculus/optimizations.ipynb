# **Optimization of Square Loss**

### Definition of Squared Loss Function

The squared loss for a parameter $x$ is:

$$
L(x) = \frac{1}{n}\sum_{i=1}^{n}(x-a_i)^2
$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.array([2, 4, 5, 7, 9])
```

```python
xs = np.linspace(0, 10, 200)
losses = [np.sum(np.square(x-a)) for x in xs]

plt.grid()
plt.xlabel("x")
plt.ylabel("Loss")
plt.title("Squared Loss Function")
plt.plot(xs, losses)
```

## Minimizing the squared loss

```python
mean = np.sum(a) / len(a)

plt.grid()
plt.plot(xs, losses)
plt.scatter(mean, np.sum(np.square(mean - a)))
plt.xlabel("x")
plt.ylabel("Loss")
plt.title("Minimum Squared Loss")
plt.show()
```

### Visualizing minimum loss arround data points

```python
x = 5
errors = (x - a)**2

plt.bar(a, errors)
plt.xlabel("Data points")
plt.ylabel("Squared Error")
plt.title(f"Squared Errors when x = {x}")
plt.show()
```
