# **Optimization Using Gradient Descent in One Variable**

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Defining $f(x)$ and $f'(x)$

```python
def f(x):
    return np.exp(x) - np.log(x)

def dfdx(x):
    return np.exp(x) - 1/x
```

### Visualizing the function

```python
xs = np.linspace(0.01, 5, 1000)
ys = f(xs)

plt.grid()
plt.title("$f(x)=e^x - log(x)$")
plt.plot(xs, ys)
```

### Gradient Descent Function

```python
def gradient_descent(initial_point, dfdx, learning_rate=0.1, num_iterations=100):
    x = initial_point
    for _ in range(num_iterations):
        x = x - learning_rate * dfdx(x)
    return x
```

```python
gradient_descent(initial_point=1.6, dfdx=dfdx)
```

### Visualizing Steps of the Algorithm

```python
def gradient_descent_plot(initial_point, dfdx, learning_rate=0.1, num_iterations=100):
    x = initial_point

    xs = np.linspace(0.01, 5, 1000)
    ys = f(xs)

    plt.plot(xs, ys)
    plt.grid()
    plt.title("$f(x)=e^x - \\log(x)$")

    for _ in range(num_iterations):
        x = x - learning_rate * dfdx(x)
        plt.scatter(x, f(x))
    plt.scatter(x, f(x), marker="*", s=200)
    return x
```

```python
gradient_descent_plot(initial_point=1.6, dfdx=dfdx)
```
