```python
def relu(x):
    return np.maximum(0, x)
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def forward(X, W1, b1, W2, b2, W3, b3):
    Z1 = X @ W1 + b1
    A1 = relu(Z1)

    Z2 = A1 @ W2 + b2
    A2 = relu(Z2)

    Z3 = A2 @ W3 + b3
    A3 = sigmoid(Z3)

    return A3
```
