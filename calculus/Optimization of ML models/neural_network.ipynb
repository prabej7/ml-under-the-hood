# **Implementing Neural Network using Numpy**
We will be implementing 3 layered neural network for simplicity and understanding.

```python
import numpy as np
```

## **Defining Dataset**

```python
X = np.array([
    [1, 2],
    [2, 3],
    [3, 4]
])
```

```python
Y = np.array([
    [0],
    [1],
    [1]
])
```

## **Initialize Parameters**
We will be initializing weights and biases according to:
- Number of inputs
- Number of neurons
- Number of hidden layers

```python
def initialize_parameters():
    W1 = np.random.randn(2, 4)
    b1 = np.zeros((1, 4))

    W2 = np.random.randn(4, 3)
    b2 = np.zeros((1, 3))

    W3 = np.random.randn(3, 1)
    b3 = np.zeros((1, 1))

    return W1, b1, W2, b2, W3, b3
```

## **Defining Activation Functions**

```python
def relu(x):
    return np.maximum(0, x)
```

```python
def relu_derivative(x):
    return (x > 0).astype(float)
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

## **Defining Loss Function**

```python
def binary_cross_entropy(y, y_pred):
    m = y.shape[0]

    loss = -(1 / m) * np.sum(
        y * np.log(y_pred) +
        (1 - y) * np.log(1 - y_pred)
    )

    return loss
```

## **Defining Forward Pass**

```python
def forward(X, W1, b1, W2, b2, W3, b3):
    Z1 = X @ W1 + b1
    A1 = relu(Z1)

    Z2 = A1 @ W2 + b2
    A2 = relu(Z2)

    Z3 = A2 @ W3 + b3
    A3 = sigmoid(Z3)

    cache = (Z1, A1, Z2, A2, Z3, A3)

    return A3, cache
```

## **Defining Backpropagation**

```python
def backward(X, Y, cache, W2, W3):
    Z1, A1, Z2, A2, Z3, A3 = cache

    m = X.shape[0]

    dZ3 = (A3 - Y) / m
    dW3 = A2.T @ dZ3
    db3 = np.sum(dZ3, axis=0, keepdims=True)
    dA2 = dZ3 @ W3.T

    dZ2 = dA2 * relu_derivative(Z2)
    dW2 = A1.T @ dZ2
    db2 = np.sum(dZ2, axis=0, keepdims=True)
    dA1 = dZ2 @ W2.T

    dZ1 = dA1 * relu_derivative(Z1)
    dW1 = X.T @ dZ1
    db1 = np.sum(dZ1, axis=0, keepdims=True)

    return dW1, db1, dW2, db2, dW3, db3
```

## **Defining Training Loop**

```python
def train(X, Y, learning_rate=0.01, epochs=1000):
    W1, b1, W2, b2, W3, b3 = initialize_parameters()

    for epoch in range(epochs):
        A3, cache = forward(X, W1, b1, W2, b2, W3, b3)
        loss = binary_cross_entropy(Y, A3)

        dW1, db1, dW2, db2, dW3, db3 = backward(
            X, Y, cache, W2, W3
        )

        W1 -= learning_rate * dW1
        b1 -= learning_rate * db1

        W2 -= learning_rate * dW2
        b2 -= learning_rate * db2

        W3 -= learning_rate * dW3
        b3 -= learning_rate * db3

        if epoch % 100 == 0:
            print(f"Epoch {epoch}, Loss: {loss}")

    return W1, b1, W2, b2, W3, b3
```

### Training the network

```python
W1, b1, W2, b2, W3, b3 = train(X, Y)
```

## **Defining Prediction Function**

```python
def predict(X):
    A3, _ = forward(X, W1, b1, W2, b2, W3, b3)

    predictions = (A3 > 0.5).astype(int)

    return predictions
```

### Making Predictions

```python
predictions = predict(X)

print(predictions)
print(Y)
```
