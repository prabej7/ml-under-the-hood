# **Inverse Function**

$$
x \xrightarrow{f(x)} x^2 \xrightarrow{g(x)} x
$$

Here, \(g(x)\) and \(f(x)\) are **inverse functions** of each other:

$$
g(f(x)) = x
$$

and

$$
f(g(x)) = x
$$


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 5, 1000)
y = np.square(x)

fig, axes = plt.subplots(1, 2)

axes[0].set_title("f(x) = x²")
axes[0].set_xlabel("x")
axes[0].set_ylabel("f(x)")
axes[0].grid()
axes[0].plot(x, y)

x = np.linspace(0, 5, 1000)
y = np.sqrt(x)

axes[1].set_title("g(x) = √x")
axes[1].set_xlabel("x")
axes[1].set_ylabel("g(x)")
axes[1].grid()
axes[1].plot(x, y)

plt.tight_layout()
plt.show()
```

## Example

Let's take:

$$
x = \frac{3}{2}
$$

Then:

$$
y = f\left(\frac{3}{2}\right)
= \left(\frac{3}{2}\right)^2
= \frac{9}{4}
$$

For the inverse function, we swap \(x\) and \(y\):

$$
x = \frac{9}{4}
$$

$$
y = g\left(\frac{9}{4}\right)
= \sqrt{\frac{9}{4}}
= \frac{3}{2}
$$

Therefore:

$$
g(f(x)) = x
$$

## **Derivative of the Inverse Function**

Consider the points

$$
P = \left(\frac{3}{2}, \frac{9}{4}\right)
$$

on $f(x) = x^2$, and

$$
Q = \left(\frac{9}{4}, \frac{3}{2}\right)
$$

on $g(x) = \sqrt{x}$.

These points demonstrate the relationship between the derivatives of inverse functions.

### 1. Slope of $f(x) = x^2$

The derivative is

$$
f'(x) = 2x
$$

At $x = \frac{3}{2}$:

$$
f'\left(\frac{3}{2}\right)
= 2\left(\frac{3}{2}\right)
= 3
$$

Therefore, the slope at $P$ is

$$
\boxed{3}
$$

### 2. Slope of the inverse $g(x) = \sqrt{x}$

Since an inverse function swaps the $x$ and $y$ coordinates, the corresponding point is

$$
Q = \left(\frac{9}{4}, \frac{3}{2}\right)
$$

The derivative of $g(x)$ is

$$
g'(x) = \frac{1}{2\sqrt{x}}
$$

At $x = \frac{9}{4}$:

$$
g'\left(\frac{9}{4}\right)
= \frac{1}{2\sqrt{\frac{9}{4}}}
= \frac{1}{2 \cdot \frac{3}{2}}
= \frac{1}{3}
$$

Therefore, the slope at $Q$ is

$$
\boxed{\frac{1}{3}}
$$

Notice that the slopes are reciprocals:

$$
3 \cdot \frac{1}{3} = 1
$$

Therefore,

$$
g'(f(x)) = \frac{1}{f'(x)}
$$

### 3. Derivative of the Inverse Function

More generally, if

$$
y = f(x)
$$

and

$$
g = f^{-1},
$$

then

$$
\boxed{g'(y) = \frac{1}{f'(x)}}
$$

Since $y = f(x)$, we can write this as

$$
\boxed{(f^{-1})'(f(x)) = \frac{1}{f'(x)}}
$$

This is the **derivative of the inverse function** formula.
