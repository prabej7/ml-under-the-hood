# **Derivative of Log(x)**


## **1. Exponential and logarithmic functions**

The exponential function is:

$$
f(x) = e^x
$$

and its derivative is:

$$
f'(x) = e^x
$$

The natural logarithm is the inverse of the exponential function:

$$
f^{-1}(x) = \log(x)
$$

This means:

$$
e^{\log(x)} = x
$$

and:

$$
\log(e^x) = x
$$

## **2. Visualizing the functions**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
fig, axes = plt.subplots(1, 2)

x = np.linspace(1, 5, 1000)
y = np.exp(x)

axes[0].grid()
axes[0].set_title("$f(x) = e^x$")
axes[0].plot(x, y)

y = np.log(x)
axes[1].grid()
axes[1].set_title("$f^-1(y) = log(y)$")
axes[1].plot(x, y)
```

## **3. Slope of the inverse function**

We know that:

$$
f(x) = e^x
$$

has derivative:

$$
f'(x) = e^x
$$

Since $\log(x)$ is the inverse of $e^x$, the slope of the inverse function is the reciprocal of the slope of the original function.

For inverse functions:
$$
\frac{1}{f'(f^{-1}(x))}
$$

Since:

$$
f^{-1}(x) = \log(x)
$$

we get:

$$
\frac{1}{e^{\log(x)}}
$$

But:

$$
e^{\log(x)} = x
$$

Therefore:

$$
\boxed{
\frac{d}{dx}\log(x) = \frac{1}{x}
}
$$
