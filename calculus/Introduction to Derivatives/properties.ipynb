# **Properties of derivatives**

## 1. Multiplication by a scalar

The derivative of the function: 
$$f = cg$$ is $$f' = cg'$$

---

Taking a function $y = x^2$ and multiply the function with a scalar $2$ we get: $y = 2x^2$.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def line1(x):
    return x**2

def line2(x):
    return 2 * line1(x)
```

```python
fig, axes = plt.subplots(1,2)

x = np.linspace(-5, 5, 1000)
y = line1(x)

axes[0].set_title("$y=x^2$")
axes[0].grid()
axes[0].plot(x,y)

y = line2(x)

axes[1].set_title("$y=2x^2$")
axes[1].grid()
axes[1].plot(x,y)
```

```python
x = np.linspace(-5, 5, 1000)
y = line1(x)

plt.plot(x,y, label="$y=x^2$")

y = line2(x)

plt.grid()
plt.plot(x,y, label="$y=2x^2$")
plt.legend()
```

```python
def compute_slope(line):
    x = np.linspace(-5, 5, 1000)
    y = line(x)
    
    point1 = (x[0], y[0])
    point2 = (x[1], y[1])
    
    delta_y = point2[1] - point1[1]
    delta_x = point2[0] - point1[0]
    
    slope = delta_y / delta_x
    return slope
```

```python
slope1 = compute_slope(line1)
slope2 = compute_slope(line2)

print("Slope for the points of line1: ",slope1)
print("Slope for the points of line2: ",slope2)
print("Since line2 is 2 times the line1:", slope2 / slope1)
```

### Conclusion

The slope of `line2` is exactly **2 times** the slope of `line1`.

This is because:

- `line1(x) = x²`
- `line2(x) = 2x²`

Therefore, multiplying a function by a constant also multiplies its rate of change (slope) by the same constant.

$$
[
\frac{\Delta y_2}{\Delta x} = 2\frac{\Delta y_1}{\Delta x}
]
$$

From our calculation:

- Slope of `line1` ≈ **-9.99**
- Slope of `line2` ≈ **-19.98**
- Ratio = **2.0**

Hence, the result confirms that **scaling the output of a function by 2 scales its slope by 2**.

## 2. Sum Rule

The **sum rule** states that the derivative of the sum of two functions is the sum of their derivatives.

$$
\frac{d}{dx}[f(x) + g(x)]
=
\frac{df(x)}{dx} + \frac{dg(x)}{dx}
$$

For example, if:

$$
f(x) = 2x
$$

$$
g(x) = 3x
$$

Then:

$$
f'(x) = 2
$$

$$
g'(x) = 3
$$

Therefore:

$$
(f(x) + g(x))' = f'(x) + g'(x)
$$

$$
= 2 + 3 = 5
$$

So, the slope of the sum of two lines is the sum of their individual slopes.

```python
def line1(x):
    return 2 * x + 3

def line2(x):
    return 3 * x + 2

def line3(x):
    return line1(x) + line2(x)
```

```python
x = np.linspace(-5, 5, 1000)
y = line1(x)

plt.grid()
plt.plot(x,y, label="$y=2x+3$")

y = line2(x)

plt.grid()
plt.plot(x,y, label="$y=3x+2$")

y = line3(x)

plt.grid()
plt.plot(x,y, label="$y=5x+5$")

plt.legend()
```

```python
slope1 = compute_slope(line1)
slope2 = compute_slope(line2)

slope3 = compute_slope(line3)
```

```python
print("Slope of line1:", slope1)
print("Slope of line2:", slope2)
print("Slope of line3 (sum of line1 and line2):", slope3)
```

### Conclusion

The experiment verifies the **sum rule of derivatives**:

$$
(f(x) + g(x))' = f'(x) + g'(x)
$$

The slopes were:

- Line 1: $2$
- Line 2: $3$
- Sum of Line 1 and Line 2: $5$

Since:

$$
2 + 3 = 5
$$

the slope of the sum is equal to the sum of the individual slopes. This demonstrates that differentiation is **linear with respect to addition**.

## 3. Product Rule
The derivative of $f = gh$ is $f' = g'h + gh'$.

```python
def line1(x):
    return 2 * x + 3

def line2(x):
    return 3 * x + 2

def line3(x):
    return line1(x) * line2(x)
```

```python
x = np.linspace(-5, 5, 1000)
y = line1(x)

plt.grid()
plt.plot(x,y, label="$y=2x+3$")

y = line2(x)

plt.grid()
plt.plot(x,y, label="$y=3x+2$")

y = line3(x)

plt.grid()
plt.plot(x,y, label="$y = 6x^2 + 9x + 6$")

plt.legend()
```

```python
def compute_slope(line, x):
    h = 0.0001
    
    y1 = line(x)
    y2 = line(x + h)
    
    delta_y = y2 - y1
    delta_x = h
    
    slope = delta_y / delta_x
    return slope

x = 2

slope1 = compute_slope(line1, x)
slope2 = compute_slope(line2, x)
slope3 = compute_slope(line3, x)
```

```python
x = 2

product_rule = slope1 * line2(x) + line1(x) * slope2

print("Slope of line3:", slope3)
print("Product rule:", product_rule)
```

### Conclusion

The experiment verifies the **product rule of derivatives**:

$$
\frac{d}{dx}[g(x)h(x)]
=
g'(x)h(x) + g(x)h'(x)
$$

At a given point $x$, the derivative of the product is not simply the product of the two derivatives. Instead, it is the sum of the two terms:

$$
g'(x)h(x) + g(x)h'(x)
$$

This demonstrates that when two functions are multiplied, both functions contribute to the rate of change of their product.

## **4. Chain Rule**

The derivative of: $$g(h(t))$$ is $\frac{d}{dt}$ $g(h(t))$ = $$\frac{dg}{dh} . \frac{dh}{dt}$$

$f(g(h(t)))$ -> $f'(g(h(t))).g'(h(t)).h'(t)$
