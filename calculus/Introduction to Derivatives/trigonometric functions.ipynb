# **Derivative of Trigonometric Function**

```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
x = np.arange(-360, 361, 1)

y = np.sin(np.radians(x))
plt.plot(x, y, label="sin(x)")

y = np.cos(np.radians(x))
plt.plot(x, y, label="cos(x)")

plt.xticks(np.arange(-360, 361, 90))
plt.grid()
plt.title("sin(x) vs cos(x)")
plt.xlabel("Degree")
plt.ylabel("Value")
plt.legend()

plt.tight_layout()
plt.show()
```

### **Sine Function**

$$
f(x) = \sin(x)
$$

The derivative of $\sin(x)$ is:

$$
f'(x) = \cos(x)
$$

### **Cosine Function**

$$
g(x) = \cos(x)
$$

The derivative of $\cos(x)$ is:

$$
g'(x) = -\sin(x)
$$

Therefore:

$$
\boxed{
\frac{d}{dx}\sin(x) = \cos(x)
}
$$

$$
\boxed{
\frac{d}{dx}\cos(x) = -\sin(x)
}
$$

---

## **Trigonometric Values and Slopes**

Since

$$
\frac{d}{dx}\sin(x) = \cos(x)
$$

the slope of $\sin(x)$ at an angle $x$ is $\cos(x)$.

And since

$$
\frac{d}{dx}\cos(x) = -\sin(x)
$$

the slope of $\cos(x)$ at an angle $x$ is $-\sin(x)$.

| Angle $x$ | $\sin(x)$ | Slope of $\sin(x)$ = $\cos(x)$ | $\cos(x)$ | Slope of $\cos(x)$ = $-\sin(x)$ |
|----------:|----------:|--------------------------------:|----------:|----------------------------------:|
| $0^\circ$   | $0$    | $1$     | $1$    | $0$     |
| $30^\circ$  | $0.5$  | $0.866$ | $0.866$| $-0.5$  |
| $45^\circ$  | $0.707$| $0.707$ | $0.707$| $-0.707$|
| $60^\circ$  | $0.866$| $0.5$   | $0.5$  | $-0.866$|
| $90^\circ$  | $1$    | $0$     | $0$    | $-1$    |
| $180^\circ$ | $0$    | $-1$    | $-1$   | $0$     |
| $270^\circ$ | $-1$   | $0$     | $0$    | $1$     |
| $360^\circ$ | $0$    | $1$     | $1$    | $0$     |

### Key observation

The value of one trigonometric function determines the slope of the other:

$$
\boxed{\text{slope of }\sin(x)=\cos(x)}
$$

$$
\boxed{\text{slope of }\cos(x)=-\sin(x)}
$$
