# **Derivative of the Exponential Function ($e^x$)**

The number $e$ can be defined using the limit:

$$
e = \lim_{n\to\infty}\left(1+\frac{1}{n}\right)^n
$$

Numerically,

$$
e \approx 2.71828
$$

The exponential function is:

$$
f(x)=e^x
$$

Its remarkable property is that its rate of change is equal to its current value:

$$
\boxed{\frac{d}{dx}e^x=e^x}
$$

Therefore,

$$
f(x)=e^x
$$

and

$$
f'(x)=e^x
$$

So the function and its derivative are the same.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(1, 5, 1000)
y = np.exp(x)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("$e^x$")
plt.title("Exponential Function")
plt.grid()
plt.show()
```
