# **Optimization Using Gradient Descent Two Variable**

### Defining the function

```python
def f(x, y):
    return 85 - (1/90) * (x**2) *  (x - 6) * (y**2) * (y - 6)
```

### Calculating the partial derivatives

```python
def dfdx(x, y):
    return (-1/90) * x * (3*x - 12) * (y**2) * (y - 6)

def dfdy(x, y):
    return (-1/90) * (x**2) * (x - 6) * y * (3*y - 12)
```

### Defining Gradient Descent Function

```python
def gradient_descent(initial_point, dfdx, dfdy, learning_rate=0.1, num_iterations=100):
    x, y = initial_point
    history = [(x, y)]
    for _ in range(num_iterations):
        x, y = x - learning_rate * dfdx(x,y), y - learning_rate * dfdy(x,y)
        history.append((x, y))
    return history
```

### Calculating the Steps and Minimum

```python
num_iterations = 30; learning_rate = 0.25; x_initial = 0.5; y_initial = 0.6
history = gradient_descent(
    initial_point=(x_initial, y_initial), 
    dfdx=dfdx, dfdy=dfdy, 
    learning_rate=learning_rate, 
    num_iterations=num_iterations
)

print("Optimal Minimal: ",history[-1])
```

### Visualizing the Steps

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
xs, ys = zip(*history)

margin = 1

x_min, x_max = min(xs) - margin, max(xs) + margin
y_min, y_max = min(ys) - margin, max(ys) + margin

X, Y = np.meshgrid(
    np.linspace(x_min, x_max, 100),
    np.linspace(y_min, y_max, 100)
)

Z = f(X, Y)

plt.figure(figsize=(6, 4))

plt.contour(X, Y, Z, levels=20)

plt.plot(xs, ys, marker="o", markersize=3)

plt.scatter(xs[0], ys[0], s=100, label="Start")
plt.scatter(xs[-1], ys[-1], s=100, label="Minimum")

plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()

plt.show()
```
