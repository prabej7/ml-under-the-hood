# **Derivatives**
- A derivative is the instantaneous rate of change of a function.
- Geometrically, it is the slope of the tangent line to the function at a particular point.

## **Slope**

The general formula for the slope between two points $(x_1, y_1)$ and $(x_2, y_2)$:

$$
m = \frac{y_2-y_1}{x_2-x_1}
$$

Or equivalently:

$$
m = \frac{\Delta y}{\Delta x}
$$

Where:

* \(m\) = slope
* \(\Delta y\) = change in \(y\)
* \(\Delta x\) = change in \(x\)


## **Some common derivatives**
### 1. Derivative of a constant

$$
y = f(x) = c
$$

$$
f'(x) = 0
$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-5, 5, 5)
y = np.array([3] * 5)
plt.grid()
plt.plot(x, y)
```

### 2. Derivative of a line

$$
y = f(x) = ax+b
$$

$$
f'(x) = a
$$

Let take a line:
$$
f(x) = 2x + 3
$$

```python
def line(x):
    return 2 * x + 3

x = np.linspace(-5, 5, 100)
y = line(x)

plt.grid()
plt.plot(x, y)
```

### 3. Derivative of Quadratic Function

$$
y = f(x) = x^2
$$

$$
f'(x) = 2x
$$

The derivative is no longer constant. It changes depending on $x$.

For example:

$$
f'(1) = 2
$$

$$
f'(3) = 6
$$

$$
f'(-2) = -4
$$

This means the slope of the curve is different at every point.

```python
x = np.linspace(-5,5,1000)
y = x**2

plt.grid()
plt.plot(x, y)
```

### 4. Derivative of Cubic Function

$$
y = f(x) = x^3
$$

$$
f'(x) = 3x^2
$$

```python
x = np.linspace(-5,5,1000)
y = x**3

plt.grid()
plt.plot(x,y)
```

### 5. Derivative of Reciprocal Function

$$
y = f(x) = \frac{1}{x}
$$

$$
f'(x) = -\frac{1}{x^2}
$$

```python
x = np.linspace(-5,5,1000)
y = 1/x

plt.grid()
plt.plot(x,y)
```
