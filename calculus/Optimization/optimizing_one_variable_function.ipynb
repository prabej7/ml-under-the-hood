# **Optimizing Functions of One Variable: Cost Minimization**

```python
import numpy as np
```

### 1. Defining Dataset

```python
prices_A = np.array([104, 108, 101, 104, 102], dtype=np.float32)
prices_B = np.array([76, 76, 84, 79, 81], dtype=np.float32)
```

### 2. Defining the function

```python
def f(omega):
    return prices_A * omega + prices_B * (1 - omega)
```

### 3. Defining the loss function

```python
def loss(omega):
    return 1/len(f(omega)) * np.sum((f(omega) - np.mean(f(omega)))**2)
```

### 4. Helper function to compute losses for respective candidate solutions

```python
def compute_losses(omega_array):
    losses = []
    for omega in omega_array:
        losses.append(loss(omega))
    return np.array(losses)
```

### 5. Preparing candidate solutions and computing losses

```python
omega_array = np.linspace(0, 1, 1001)
losses = compute_losses(omega_array)
```

### 6. Conclusion
We got the optimal solution for the function.

```python
optimal_idx = losses.argmin()
optimal_omega = omega_array[optimal_idx]
loss = losses[optimal_idx]

print("Optimal Solution: ", optimal_omega)
print("Loss: ", loss)
```
