# **Rank of a Matrix**

```python
import numpy as np
```

```python
A = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [1, 1, 2]
], dtype=float)
```

```python
A
```

```python
def gaussian_elimination(A, b=None):
    A = np.asarray(A, dtype=np.float64)

    if b is not None:
        b = np.asarray(b, dtype=np.float64)
        mat = np.hstack((A, b.reshape(-1, 1)))
    else:
        mat = A.copy()

    def eliminate(pivot_index):
        for index, row in enumerate(mat):
            if index == pivot_index:
                continue

            multiplier = row[pivot_index]
            mat[index] = row - (mat[pivot_index] * multiplier)

    for i in range(len(mat)):
        if mat[i][i] == 0:
            for j in range(i + 1, len(mat)):
                if mat[j][i] != 0:
                    mat[i], mat[j] = mat[j].copy(), mat[i].copy()
                    break

        pivot_element = mat[i][i]

        if pivot_element == 0:
            continue

        mat[i] = mat[i] / pivot_element

        eliminate(i)

    return mat
```

```python
def matrix_rank(A):
    rref = gaussian_elimination(A)
    
    rank = 0

    for row in rref:
        # If the row contains any non-zero value,
        # then it contains a pivot
        if np.any(row != 0):
            rank += 1

    return rank
```

```python
matrix_rank(A)
```

```python
np.linalg.matrix_rank(A)
```

```python
v1 = np.array([1, 2, 3], dtype=float)
v2 = np.array([2, 4, 6], dtype=float)
v3 = np.array([1, 1, 1], dtype=float)
```

```python
def is_linearly_independent(vectors):
    vecs = tuple(vectors)
    A = np.vstack(vecs)
    rank = matrix_rank(A)
    return rank == len(A)
```

```python
is_linearly_independent([v1, v2, v3])
```
