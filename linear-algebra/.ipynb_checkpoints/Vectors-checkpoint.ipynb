```python
import numpy as np
import math
```

```python
v1 = np.array([2, 3])
v2 = np.array([4, 1])
```

```python
def dot_product(v1, v2):
    n = len(v1)
    if len(v2) != n:
        raise ValueError("The lenght of vectors must be equal")
    res = 0
    for i in range(n):
        res += v1[i] * v2[i]

    return res
```

```python
dot_product(v1, v2)
```

```python
def direction(v):
    return math.atan(v[1]/v[0])
```

```python
direction(v1)
```

# **Matrix as Linear Transformation**

```python
A = np.array([
    [2, 0],
    [0, 3]
])

v = np.array([4, 2])
```

```python
def transform(A, v):
    tranformation = [0] * len(A)
    for i, row in enumerate(A):
        tranformation[i] = dot_product(row, v)
    return tranformation
```

```python
transform(A, v)
```

```python
A = np.array([
    [0, -1],
    [1,  0]
])

v = np.array([2, 3])

transform(A, v)
```

```python
def mat_mul(A, B):
    C = [[0] * len(B[0]) for _ in range(len(A))]

    for r in range(len(A)):
        for c in range(len(B[0])):
            dot_prd = dot_product(A[r], B[:, c])
            C[r][c] = dot_prd

    return np.array(C)
```

```python
A = np.array([
    [3, 1, 4],
    [2, -1, 2]
])

B = np.array([
    [3, 0, 1, -2],
    [1, 5, 2, 0],
    [-2, 1, 4, 0]
])
```

```python
mat_mul(A, B)
```

```python
def gaussian_elimination(A, b):
    mat = np.hstack((A, b.reshape(-1,1)), dtype=np.float64)

    def eliminate(pivot_index):
        for index, row in enumerate(mat):
            if index == pivot_index:
                continue
            multiplier = row[pivot_index]
            row = row - (mat[pivot_index] * multiplier)
            mat[index] = row

    for i in range(len(mat)):
        if mat[i][i] == 0:
            for j in range(i + 1, len(mat)):
                if mat[j][i] != 0:
                    mat[i], mat[j] = mat[j].copy(), mat[i].copy() 
                    break
        pivot_element = mat[i][i]
        if pivot_element == 0:
            continue

        mat[i] = mat[i] / pivot_element

        eliminate(i)
    return np.array(mat)
```

```python
def inverse(A):
    C = np.eye(len(A))
    B = [[0] * len(C[0]) for _ in range(len(C))]
    
    for i in range(len(C)):
        ans = gaussian_elimination(A, C[:, i])[:, -1] # since the last column has the solution
        for index, an in enumerate(ans):
            B[index][i] = an
    return np.array(B)
```

```python
A = np.array([
    [3, 1],
    [1, 2]
])

C = np.array([
    [1, 0],
    [0, 1]
])

inverse(A)
```

```python
mat_mul(A, inverse(A))
```
