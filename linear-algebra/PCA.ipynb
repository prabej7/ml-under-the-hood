# **Implementing PCA using Numpy**

```python
import numpy as np
```

#### 1. Helper function to create mean matrix

```python
def compute_mean_matrix(A):
    mean = []
    n = len(A)
    for i in range(len(A[0])):
        mean.append(np.mean(A[:,i]))
    return np.array([mean for _ in range(n)])
```

#### 2. Helper function to create covariance matrix

```python
def convariance_matrix(A):
    n = len(A)    
    mean_matrix = compute_mean_matrix(A)
    A_m = A - mean_matrix
    return (np.transpose(A_m) @ A_m) / (n-1)
```

#### 3. Dataset

```python
X = np.array([
    [2, 4, 1, 5, 3],
    [3, 6, 2, 7, 4],
    [4, 8, 3, 9, 5],
    [5, 10, 4, 11, 6],
    [6, 12, 5, 13, 7]
], dtype=float)
```

#### 4. Create Covariance Matrix

```python
cov_matrix = convariance_matrix(X)
```

```python
cov_matrix
```

#### 5. Find eigen values and eigen vectors

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
```

#### 6. Sort the eigen vectors by eigen values

```python
indices = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[indices]
eigenvectors = eigenvectors[:, indices]
```

#### 7. Choose top `k` dimension for reduction

```python
k = 2
```

```python
V = eigenvectors[:, :k]
```

#### 8. Project the centered data on to the projection matrix

```python
Xpca = (X - compute_mean_matrix(X)) @ V
```

## **Compiled Function for performing PCA**

```python
def pca(A, k):
    cov_mat = convariance_matrix(A)
    eigenvalues, eigenvectors = np.linalg.eig(cov_mat)

    indices = np.argsort(eigenvalues)[::-1]
    eigenvalues = eigenvalues[indices]
    eigenvectors = eigenvectors[:, indices]

    V = eigenvectors[:, :k]
    return (A - compute_mean_matrix(A)) @ V
```
