# **Covariance Matrix**

```python
import numpy as np
```

## **Dataset**

We have $n$ data points, where each point contains two features: $x$ and $y$.

```python
A = np.array([
    [10, 5],
    [12, 3],
    [6, 9],
    [6, 4],
    [5, 11],
    [14, 2],
    [8, 1],
    [3, 13]
])
n = len(A)
```

```python
A
```

## **Mean**

We will calculate the mean of $x$ and $y$ using the following formula.

$$
\mu = \frac{1}{n}\sum_{i=1}^{n} x_i
$$

```python
m_x = np.mean(A[:,0])
m_y = np.mean(A[:,1])
```

## **Mean Centering**

We subtract the mean from each data point to center the data around zero.

$$
A_m = A - \mu
$$

```python
A_m = A - np.array([[m_x, m_y] for _ in range(n)])
```

```python
A_m
```

## **Covariance Matrix**

We calculate the covariance matrix using the centered data.

$$
C = \frac{1}{n-1}(A-\mu)^T(A-\mu)
$$

```python
cov_matrix = (np.transpose(A_m) @ A_m) / (n-1)
```

```python
cov_matrix
```

# **Covariance Matrix** - Dynamic Solution

```python
A = np.array([
    [10, 5],
    [12, 3],
    [6, 9],
    [6, 4],
    [5, 11],
    [14, 2],
    [8, 1],
    [3, 13]
])
n = len(A)
```

```python
def compute_mean_matrix(A):
    mean = []
    for i in range(len(A[0])):
        mean.append(np.mean(A[:,i]))
    return np.array([mean for _ in range(n)])
```

```python
def convariance_matrix(A):
    mean_matrix = compute_mean_matrix(A)
    A_m = A - mean_matrix
    return (np.transpose(A_m) @ A_m) / (n-1)
```

```python
convariance_matrix(A)
```
